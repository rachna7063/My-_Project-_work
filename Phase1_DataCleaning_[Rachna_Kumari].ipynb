```python
print("--- start data merging ---")

 #1. change table id in upper format

watch_history_data['title_id'] = watch_history_data['title_id'].astype(str).str.strip().str.upper()
watch_history_data['subscriber_id'] = watch_history_data['subscriber_id'].astype(str).str.strip().str.upper()

titles_data['title_id'] = titles_data['title_id'].astype(str).str.strip().str.upper()
subscribers_data['subscriber_id'] = subscribers_data['subscriber_id'].astype(str).str.strip().str.upper()

watchlist_data['subscriber_id'] = watchlist_data['subscriber_id'].astype(str).str.strip().str.upper()
watchlist_data['title_id'] = watchlist_data['title_id'].astype(str).str.strip().str.upper()

reviews_data['subscriber_id'] = reviews_data['subscriber_id'].astype(str).str.strip().str.upper()
reviews_data['title_id'] = reviews_data['title_id'].astype(str).str.strip().str.upper()

ratings_data['subscriber_id'] = ratings_data['subscriber_id'].astype(str).str.strip().str.upper()
ratings_data['title_id'] = ratings_data['title_id'].astype(str).str.strip().str.upper()

# 2. rename the clash numeric column data in Titles data
titles_data_cleaned = titles_data.rename(columns={'content_duration_min': 'title_duration_min'})

# 3. MERGING (
# Step A: Watch History + Subscriber details link
temp_df = pd.merge(watch_history_data, subscribers_data, on='subscriber_id', how='left')

# Step B: again link Title/Content catalog details
final_data = pd.merge(temp_df, titles_data_cleaned, on='title_id', how='left')

# Step C: map the Watchlist, Reviews aur Ratings
final_data = pd.merge(final_data, watchlist_data, on=['subscriber_id', 'title_id'], how='left')
final_data = pd.merge(final_data, reviews_data, on=['subscriber_id', 'title_id'], how='left')
final_data = pd.merge(final_data, ratings_data, on=['subscriber_id', 'title_id'], how='left')

# 4. Valid Flag updates for blank watchlist items
final_data['in_watchlist'] = final_data['in_watchlist'].fillna(0)
final_data['churn_date'] = final_data['churn_date'].fillna('Still Active')

print(f"✔  New Shape: {final_data.shape}")
print("\n--- rows check ---")
# country_x represents the subscriber country, country_y represents the title production country
print(final_data[['title_id', 'title_name', 'primary_genre', 'age', 'country_x']].head())
```

```python
print("--- Custom Renaming of Null Values Started ---")

# 1. Watchlist and User Activity related nulls
if 'watched' in final_data.columns:
    final_data['watched'] = final_data['watched'].fillna('Not Checked')
if 'in_watchlist' in final_data.columns:
    final_data['in_watchlist'] = final_data['in_watchlist'].fillna(0) # 0 means not in watchlist

# 2. Text/Categorical columns (Reviews and Sentiments)
text_null_cols = ['review_text', 'sentiment', 'director', 'cast', 'license_type', 'maturity_rating']
for col in text_null_cols:
    if col in final_data.columns:
        final_data[col] = final_data[col].fillna('No Feedback Given' if 'review' in col else 'Unknown')

# 3. Numeric columns (Ratings, Helpful Votes, and System Metrics)
numeric_null_cols = ['rating', 'helpful_votes', 'review_id', 'rating_id', 'quality_score', 'popularity_score', 'license_cost_usd', 'total_watch_hours', 'total_plays']
for col in numeric_null_cols:
    if col in final_data.columns:
        final_data[col] = final_data[col].fillna(0) # Setting system missing IDs and metrics to 0

# 4. Date/Time specific columns (NaT tracking)
date_null_cols = ['review_date', 'rating_date', 'license_expiry']
for col in date_null_cols:
    if col in final_data.columns:
        # Converting to string format first so we can write text instead of NaT
        final_data[col] = final_data[col].dt.strftime('%Y-%m-%d').fillna('Not Applicable')

print("✔ All Null/NaN/NaT values are successfully replaced with proper renames!")
print("\n--- Remaining Missing Counts Verification ---")
print(final_data[['watched', 'sentiment', 'review_text', 'rating', 'review_date']].isnull().sum())
```

```python
final_data.info()
```

```python
# Final clean up for the remaining watchlist tracking columns
if 'watchlist_id' in final_data.columns:
    final_data['watchlist_id'] = final_data['watchlist_id'].fillna(0)

if 'added_date' in final_data.columns:
    # Converting to string format first to write text instead of NaT tracking
    final_data['added_date'] = final_data['added_date'].dt.strftime('%Y-%m-%d').fillna('Not Applicable')

print("--- Ultimate Clean Check ---")
print(final_data[['watchlist_id', 'added_date']].isnull().sum())
```
