```python
print("--- Watch History Data Cleaning Shuru ---")

#  Convert Date column into Datetime format
watch_history_data['watch_date'] = pd.to_datetime(watch_history_data['watch_date'], errors='coerce')
```

```python
print("---check it that watch_date column converted in date time format")
watch_history_data.info()
```

**successfull**

```python
#  To clean aur standardize  the text column
text_cols = ['device', 'region']
for col in text_cols:
    watch_history_data[col] = watch_history_data[col].astype(str).str.strip().str.title()
```

```python
#  Outlier Fix: remove negative value
watch_history_data = watch_history_data[watch_history_data['watch_duration_min'] >= 0]
watch_history_data = watch_history_data[watch_history_data['content_duration_min'] > 0]
```

```python
#CHECK 1: Confirm Watch Duration is less than or equal to Content Duration
# =========================================================================
# if watch_duration > content_duration than entry (mismatch)
watch_history_data['flag_invalid_duration'] = watch_history_data['watch_duration_min'] > watch_history_data['content_duration_min']
mismatch_duration = watch_history_data['flag_invalid_duration'].sum()
print(f" Check 1: Total {mismatch_duration} rows finding where watch duration is greater than content duration.")
```

```python
#CHECK 2: TO  Recalculate the completion percentage
# =========================================================================
# TO apply formula for the right calculation
watch_history_data['completion_percentage'] = (watch_history_data['watch_duration_min'] / watch_history_data['content_duration_min']) * 100

print("\n Watch History table is completely clean aur logically validated!")

```

```python
watch_history_data['completion_percentage']
```

```python
watch_history_data['completed'] = np.where(watch_history_data['completion_pct'] >= 90, True, False)

print("--- Completion Checks Successfully Updated! ---")
print("Completion Percentage (Min):", watch_history_data['completion_pct'].min(), "%")
print("Completion Percentage (Max):", watch_history_data['completion_pct'].max(), "%")
print("\nVideo Completion Counts:")
print(watch_history_data['completed'].value_counts())
```
